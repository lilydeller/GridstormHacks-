```python
import requests
import json
import numpy as np

# Bounding box for Columbia, SC
min_lon, min_lat = -81.20, 33.90
max_lon, max_lat = -80.95, 34.10

# Resolution of sampling grid (increase for more detail)
num_points = 10

lats = np.linspace(min_lat, max_lat, num_points)
lons = np.linspace(min_lon, max_lon, num_points)

elevation_data = []

url = "https://epqs.nationalmap.gov/v1/json"

for lat in lats:
    for lon in lons:
        params = {
            "x": lon,
            "y": lat,
            "units": "Meters",
            "wkid": 4326
        }

        r = requests.get(url, params=params)
        r.raise_for_status()

        elevation = r.json()["value"]

        elevation_data.append({
            "latitude": lat,
            "longitude": lon,
            "elevation_meters": elevation
        })

print(f"Collected {len(elevation_data)} elevation points.")

with open("columbia_sc_elevation.json", "w") as f:
    json.dump({
        "location": "Columbia, SC",
        "units": "meters",
        "points": elevation_data
    }, f, indent=4)

print("Saved elevation data to columbia_sc_elevation.json")
```
