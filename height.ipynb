```python
import requests

# Bounding box for Columbia, SC (minLon, minLat, maxLon, maxLat)
bbox = "-81.20,33.90,-80.95,34.10"

url = "https://tnmaccess.nationalmap.gov/api/v1/products"

params = {
    "datasets": "Lidar Point Cloud (LPC)",
    "bbox": bbox,
    "outputFormat": "JSON",
    "max": 100
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()

print(f"Found {len(data.get('items', []))} lidar products.\n")

for item in data.get("items", []):
    print("Title:", item.get("title"))
    print("Download URL:", item.get("downloadURL"))
    print("Date:", item.get("publicationDate"))
    print("-" * 60)
```

```python
import requests
import json

bbox = "-81.20,33.90,-80.95,34.10"

url = "https://tnmaccess.nationalmap.gov/api/v1/products"

params = {
    "datasets": "Lidar Point Cloud (LPC)",
    "bbox": bbox,
    "outputFormat": "JSON",
    "max": 100
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()

with open("columbia_sc_lidar_raw.json", "w") as f:
    json.dump(data, f, indent=4)

print("Saved raw JSON to columbia_sc_lidar_raw.json")
```
